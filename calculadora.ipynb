```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
constantes_antoine = {
    "agua":    {"A": 8.07131,  "B": 1730.63,  "C": 233.426},
"etanol":  {"A": 8.20417,  "B": 1642.89,  "C": 230.300},
"benceno": {"A": 6.90565,  "B": 1211.033, "C": 220.790},
"acetona": {"A": 7.11714,  "B": 1210.595, "C": 229.664},
"tolueno": {"A": 6.95464,  "B": 1344.800, "C": 219.482}
}
componente = input("Componentes disponibles: " + str(list(constantes_antoine.keys())) + "\nIngresa un componente: ").lower()
componente2 = input("Componentes disponibles: " + str(list(constantes_antoine.keys())) + "\nIngresa un componente: ").lower()
if componente in constantes_antoine:
    print(constantes_antoine[componente])
    A = constantes_antoine[componente]["A"]
    B = constantes_antoine[componente]["B"]
    C = constantes_antoine[componente]["C"]
if componente2 in constantes_antoine:
    A2 = constantes_antoine[componente2]["A"]
    B2 = constantes_antoine[componente2]["B"]
    C2 = constantes_antoine[componente2]["C"]
    T_min = float(input("Temperatura minima(ºC):"))
    T_max = float(input("Temperatura maxima(ºC):"))
    T = np.linspace(T_min, T_max, 100)
    P = 10** (A - B/ (C + T))
    P2 = 10** (A2 - B2/ (C2+ T))
    plt.plot(T, P, linewidth=2, label=componente)
    plt.plot(T, P2, color="red", linewidth=2, label=componente2)
    plt.axhline(y=760, color="black", linestyle="--", label="P atm (760 mmHg)")
    plt.xlabel("Temperatura (°C)")
    plt.ylabel("Presión de vapor (mmHg)")
    plt.title("Ecuación de Antoine — " + componente + " vs " + componente2)
    plt.grid(True)
    plt.legend()
    plt.show()
    df = pd.DataFrame({"Temperatura": T, "Presion": P})
    df.to_csv("resultados.csv", index=False)
    print("CSV guardado")
else:
    print("componente no encontrado")
```
